```python
import pandas as pd
from pandasql import sqldf
df = pd.read_excel("/content/Dataset for Data Analytics (1).xlsx")
df.head()
```

```python
print("creating a SQL Function:")
pysqldf = lambda q: sqldf(q, globals())
```

```python
print("using SELECT Query: ")
query = """ SELECT CustomerID, Product, Totalprice FROM df """
pysqldf(query)
```

```python
print("Using WHERE Clause:")
query1 = """ SELECT * FROM df WHERE TotalPrice > 500 """
pysqldf(query1)
```

```python
print("To sort values from highest to lowest using ORDER BY:")
query2 = """ SELECT Product, TotalPrice FROM df ORDER BY TotalPrice DESC """
pysqldf(query2)
```

```python
print("Using GROUP BY:")
query3 = """ SELECT PaymentMethod, COUNT(*) as TotalOrders FROM df GROUP BY PaymentMethod """
pysqldf(query3)
```

```python
print("SUM Aggregation:")
query4 = """ SELECT Product, SUM(TotalPrice) as TotalSales FROM df GROUP BY Product """
pysqldf(query4)
```

```python
print(" Average (AVG) Aggregation:")
query5 = """ SELECT Product, AVG(TotalPrice) as AverageSales FROM df GROUP BY Product """
pysqldf(query5)
```

```python
print("using COUNT Aggregation:")
query6 = """ SELECT COUNT(*) as TotalOrders FROM df """
pysqldf(query6)
```
