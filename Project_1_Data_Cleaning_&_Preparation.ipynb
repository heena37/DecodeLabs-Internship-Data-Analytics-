```python
import pandas as pd
df = pd.read_excel("/content/Dataset for Data Analytics.xlsx")
df.info()
df.isnull().sum()
```

```python
df['CouponCode'] = df['CouponCode'].fillna('No Coupon')
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
df.duplicated().sum()
```

```python
print("Shape of the Data set after removing the duplicates:", df.shape)
```

```python
print('Filling missing values for numeric values:')
df.fillna(df.mean(numeric_only = True), inplace=True)
print(df)
```

```python
print('checking data types:')
df.dtypes
```

```python
print("save cleaned dataset:")
df.to_excel('cleaned_dataset.xlsx', index=False)
```
